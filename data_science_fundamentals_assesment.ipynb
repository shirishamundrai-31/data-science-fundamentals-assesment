```python
import pandas as pd

# Create a sample dataset
data = {
    "Age": [21, 22, 20, 23],
    "Department": ["AI", "CS", "AI", "IT"],
    "Satisfaction": ["High", "Medium", "Low", "High"],
    "Joining_Date": ["2025-01-10", "2025-02-15", "2025-03-20", "2025-04-05"],
    "Feedback": [
        "Good course",
        "Helpful teacher",
        "Need more practice",
        "Excellent learning"
    ]
}

df = pd.DataFrame(data)

# 1. Convert datetime data
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

# 2. Convert ordinal data into ordered categories
df["Satisfaction"] = pd.Categorical(
    df["Satisfaction"],
    categories=["Low", "Medium", "High"],
    ordered=True
)

# 3. Handle text data
df["Feedback"] = df["Feedback"].str.lower().str.strip()

# 4. Display the dataset
print("Dataset:")
print(df)

# 5. Display data types
print("\nData Types:")
print(df.dtypes)

# 6. Analyze numerical data
print("\nNumerical Summary:")
print(df["Age"].describe())

# 7. Analyze categorical data
print("\nDepartment Counts:")
print(df["Department"].value_counts())

# 8. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np

# 1. Reproducibility: set a random seed
np.random.seed(42)

# 2. Create a sample dataset
student_data = {
    "student_name": ["Asha", "Ravi", "Mina", "John", "Sara"],
    "study_hours": [2, 3, 1, 4, 5],
    "exam_score": [65, 75, 55, 85, np.nan]
}

df = pd.DataFrame(student_data)

# 3. Display the original dataset
print("Original Dataset:")
print(df)

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Remove duplicate records
df = df.drop_duplicates().copy()

# 6. Handle missing numerical values
df["exam_score"] = df["exam_score"].fillna(
    df["exam_score"].median()
)

# 7. Validate data
df["study_hours"] = pd.to_numeric(
    df["study_hours"], errors="coerce"
)
df["exam_score"] = pd.to_numeric(
    df["exam_score"], errors="coerce"
)

# Check valid ranges
df = df[
    df["study_hours"].between(0, 24)
    & df["exam_score"].between(0, 100)
].copy()

# 8. Calculate descriptive statistics
average_score = df["exam_score"].mean()
median_score = df["exam_score"].median()

print("\nCleaned Dataset:")
print(df)

print("\nAverage Score:", round(average_score, 2))
print("Median Score:", round(median_score, 2))

# 9. Responsible interpretation
print("\nObservation:")
print("These statistics describe only the sample dataset.")
print("They do not prove that study hours cause higher scores.")
```

```python
# DATA SCIENCE FUNDAMENTALS ASSIGNMENT

import pandas as pd
import numpy as np
import statistics
import matplotlib.pyplot as plt

# 1. Python Programming Basics
student_name = "Asha"
marks = [65, 75, 55, 85, 90]

print("Student:", student_name)
print("Marks:", marks)

# Conditional statement
for mark in marks:
    if mark >= 75:
        print(mark, "- Good Performance")
    else:
        print(mark, "- Needs Improvement")

# Function
def calculate_average(values):
    return sum(values) / len(values)

print("Average Marks:", calculate_average(marks))

# 2. Statistical Analysis
print("\nSTATISTICAL ANALYSIS")
print("Mean:", statistics.mean(marks))
print("Median:", statistics.median(marks))
print("Mode:", statistics.multimode(marks))
print("Population Variance:", statistics.pvariance(marks))
print("Population Standard Deviation:",
      round(statistics.pstdev(marks), 2))

print("75th Percentile:", np.percentile(marks, 75))

# Probability example
print("Probability of an even die result:", 3 / 6)

# Correlation
study_hours = [1, 2, 3, 4, 5]
correlation = np.corrcoef(study_hours, marks)[0, 1]
print("Correlation:", round(correlation, 2))

# 3. Create a DataFrame
data = {
    "Student": ["Asha", "Ravi", "Mina", "John", "Sara"],
    "Study_Hours": study_hours,
    "Marks": marks,
    "Department": ["AI", "CS", "AI", "CS", "AI"]
}

df = pd.DataFrame(data)

print("\nDATASET")
print(df)

# 4. Data Types
print("\nDATA TYPES")
print(df.dtypes)

# 5. Data Manipulation
print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStudents scoring 75 or above")
print(df[df["Marks"] >= 75])

print("\nAverage Marks by Department")
print(df.groupby("Department")["Marks"].mean())

print("\nSorted by Marks")
print(df.sort_values("Marks", ascending=False))

# 6. Data Visualization
plt.figure(figsize=(8, 4))
plt.bar(df["Student"], df["Marks"])
plt.title("Student Performance Analysis")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.show()

# 7. Save the dataset and results
df.to_csv("student_performance.csv", index=False)

print("\nDataset saved as student_performance.csv")

# 8. Conclusion
print("\nCONCLUSION")
print("The dataset contains", len(df), "students.")
print("The average mark is", round(df["Marks"].mean(), 2))
print("This analysis describes the sample only.")
print("Correlation does not prove causation.")
```
